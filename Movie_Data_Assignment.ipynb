### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
# Gross revenue ko numeric mein convert karna
data["Gross"] = pd.to_numeric(
    data["Gross"].astype(str).str.replace(",", ""),
    errors="coerce"
)

# Drama aur Action movies filter karna
drama = data[
    data["Genre"].str.contains("Drama", na=False)
]["Gross"].dropna()

action = data[
    data["Genre"].str.contains("Action", na=False)
]["Gross"].dropna()

# Box Plot
mt.figure(figsize=(10, 6))

mt.boxplot(
    [drama, action],
    tick_labels=["Drama", "Action"]
)

mt.title("Gross Revenue Comparison: Drama vs Action Movies")
mt.xlabel("Genre")
mt.ylabel("Gross Revenue (USD)")
mt.grid(axis="y", alpha=0.3)

mt.tight_layout()
mt.show()
```

### **Question 38**

```python
# Gross ko numeric format mein convert karna
data["Gross"] = pd.to_numeric(
    data["Gross"].astype(str).str.replace(",", ""),
    errors="coerce"
)

# Pooray dataset ka median Gross
median_gross = data["Gross"].median()

# Movies filter karna
movies = data[
    (pd.to_numeric(data["Released_Year"], errors="coerce") < 2000) &
    (data["Gross"] > median_gross)
]

# Movie names aur ratings display karna
print(movies[["Series_Title", "IMDB_Rating"]])

print("Overall Median Gross:", median_gross)
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import numpy as np

# Top 5 highest-rated movies
top5 = data.sort_values(
    by="IMDB_Rating",
    ascending=False
).head(5).copy()

# IMDb rating ko 100 ke scale par convert karna
top5["IMDB_Scaled"] = top5["IMDB_Rating"] * 10

# Missing Meta_score wali movies remove karna
top5 = top5.dropna(subset=["Meta_score"])

x = np.arange(len(top5))
width = 0.35

mt.figure(figsize=(12, 6))

mt.bar(
    x - width / 2,
    top5["Meta_score"],
    width,
    label="Meta Score"
)

mt.bar(
    x + width / 2,
    top5["IMDB_Scaled"],
    width,
    label="IMDb Rating (Scaled to 100)"
)

mt.title("Top 5 Highest Rated Movies: Meta Score vs IMDb Rating")
mt.xlabel("Movie Name")
mt.ylabel("Score out of 100")
mt.xticks(x, top5["Series_Title"], rotation=30, ha="right")
mt.ylim(0, 100)
mt.legend()
mt.grid(axis="y", alpha=0.3)
mt.tight_layout()
mt.show()
```
