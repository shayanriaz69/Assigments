**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        self.__balance = self.__balance + amount
        return f"Your New Balance : {self.__balance}"
    def withdraw(self,request):
        if request <= self.__balance:
            self.__balance = self.__balance - request
            print("Withdraw Successfully")
            return f"Your Remaining Balance : {self.__balance}"
        else:
            print("Not Enough Balance")

b1 = BankAccount(1000)
```

```python
b1.withdraw(1100)
```

```python
b1.deposit(4500)
```

```python
# Write your code here for Q6
class Vehicle:
    def __init__(self,brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self,brand,model):
        super().__init__(brand)
        self.model = model
        
    def show_info(self):
        print(f"Car Brand Name is {self.brand} and Model {self.model}")

car1 = Car('Toyota','2026')
car1.show_info()
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
# Write your code here for Q9
class parent:
    def __init__(self,father,mother):
        self.father = father
        self.mother = mother
    def travelling(self):
        print("Parent Are Going Abroad")

class Child(parent):
    def __init__(self,father,mother):
        super().__init__(father,mother)
    def show_info(self):
        print(f"Father Name {self.father} and Mother Name {self.mother}")

c = Child("imran","Rubina")
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class Shape:
    def show_shape(self):
        print("Shape")

class Circle(Shape):
    def show_circle(self):
        print("Circle Hierarchichal Interitance")

class Square(Shape):
    def show_square(self):
        print("Square Hierarchichal Inheritance")

cir = Circle()
sq = Square()
cir.show_shape()
cir.show_circle()
sq.show_shape()
sq.show_square()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13

class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


n1 = Number(99)
n2 = Number(31)

n1 + n2
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18

from abc import ABC, abstractmethod

class Car(ABC):

    def start_engine(self):
        print("Engine Started")

    @abstractmethod
    def drive(self):
        pass


class Toyota(Car):
    def drive(self):
        print("Car Is Driving")


car1 = Toyota()

car1.start_engine()
car1.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Credit Card")


class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using UPI")


p1 = CreditCardPayment()
p2 = UPIPayment()

p1.pay(5000)
p2.pay(2000)
```
